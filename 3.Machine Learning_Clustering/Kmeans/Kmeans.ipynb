```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
dataset=pd.read_csv("Mall_Customers.csv")
dataset
```

```python
x=dataset.iloc[:,3:5].values  # x getting input from dataset of rows, cloumns 3 and 4
```

```python
x
```

```python
#create a empty list to store the kmeans value. init='k-means++' is initially cenroid the center value by using for loop by iteration it will find out the number of clusters.
from sklearn.cluster import KMeans       
list1=[]
for i in range(1,11):
    Kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    Kmeans.fit(x)
    list1.append(Kmeans.inertia_)
plt.plot(range(1,11),list1)
plt.title("The Elbow Method")
plt.xlabel("The Number of clusters")
plt.ylabel("wcss")
plt.show
```

```python
from sklearn.cluster import KMeans
kmeans=KMeans(n_clusters=5,init='k-means++',random_state=42)
y_Kmeans=Kmeans.fit_predict(x)
```

```python
y_Kmeans
```

```python
supervised=dataset
supervised
```

```python
supervised["cluster_group"]=y_Kmeans
```

```python
supervised
```

```python
supervised.to_csv("cluster.csv",index=False)
```
