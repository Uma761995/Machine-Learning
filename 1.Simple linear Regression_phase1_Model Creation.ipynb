```python
# pandas is a library contains the function of the database
import pandas as pd 
```

```python
dataset=pd.read_csv("Salary_Data.csv") # dataset read the database. here .csv is comma seperated value 
```

```python
dataset # All databases are stored into the dataset
```

```python
#input and output split
```

```python
#from the database seperating the input data and to get seperate column from the table use [[" "]]
independent=dataset[["YearsExperience"]] 
independent
```

```python
# from the database seperating the output data
dependent=dataset[["Salary"]]
dependent
```

```python
# sklearn is a library supports in Machine Learning here selecting the model from sklearn to import train_test_split 
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,y_test=train_test_split(independent, dependent, test_size=0.30, random_state=0) 
# seperating the actual input and actual output parameter in x_train and Y_train.
# X_test and y_test assigning for user given input and predicted output.
```

```python
X_train,X_test,Y_train,y_test
```

```python
# Model creation
```

```python
# from linear modelz importing linear regression
from sklearn.linear_model import LinearRegression
regressor=LinearRegression() # regressor assigning the linearregression as a class type
regressor.fit(X_train,Y_train) # regressor access to fit the acutal input and output once it get accessed the slope(weight W) and the bais (b) value had been calculated once it done the model is created
```

```python
weight=regressor.coef_    #weight to check the slope value of the database by using coef_ function
weight
```

```python
bais=regressor.intercept_  #bais to check the origin or intercept or initial value of the database by using intercept_
bais
```

```python
Y_pred=regressor.predict(X_test)  #To cross check predicted output from x_test
```

```python
# Validating metrics
```

```python
from sklearn.metrics import r2_score  #by using r2_score from metrics method validating the model which is created
r_score=r2_score(y_test,Y_pred)
```

```python
r_score  # the result shows nearly to 1 the model is good if it is nearly to 0 the model is poor
```

```python
import pickle  # pickle is the function name to save the model and the exension is .sav it is mandatory
filename="finalized_model_linear.sav"  # filname created
```

```python
pickle.dump(regressor,open(filename,'wb'))  #file created, wb is write binary
```

```python
loaded_model=pickle.load(open("finalized_model_linear.sav",'rb'))  #loaded model assigning to load the file to get predicted output, rb is read binary
result=loaded_model.predict([[15]]) #finally the result shows the predicted output for user given input
```
