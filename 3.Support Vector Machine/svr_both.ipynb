```python
import pandas as pd
```

```python
dataset=pd.read_csv("50_Startups.csv")
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend', 'State_Florida', 'State_New York']]
dependent=dataset[["Profit"]]
from sklearn.model_selection import train_test_split
(x_train,x_test,y_train,y_test)=train_test_split(independent,dependent, test_size=0.30, random_state=0)
```

```python
x_train
```

```python
from sklearn.preprocessing import StandardScaler 
sc=StandardScaler()
x_train=sc.fit_transform(x_train)    
x_test=sc.transform(x_test)
```

```python
scy=StandardScaler()
y_train=scy.fit_transform(y_train)
y_test=scy.transform(y_test)
```

```python
from sklearn.svm import SVR  
regressor=SVR(kernel="linear",C=1000)  
regressor.fit(x_train,y_train)
```

```python
regressor.intercept_
```

```python
regressor.n_support_
```

```python
regressor.support_
```

```python
y_pred=regressor.predict(x_test)
```

```python
x_test
```

```python
from sklearn.metrics import r2_score  #validation metrics
r_score=r2_score(y_test,y_pred)
r_score 
```

```python
import pickle
filename="finalized_model_svr_both.sav"
scale_input="Scaled_data.sav"
scale_output="scaled_output.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
pickle.dump(sc,open(scale_input,'wb'))
pickle.dump(scy,open(scale_output,'wb'))
```
