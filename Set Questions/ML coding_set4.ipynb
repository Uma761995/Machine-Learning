```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
data={'Name': ['Arjun', 'Leela', 'Kiran'], 'Gender': ['Male', 'Female', 'Male']}
df=pd.DataFrame(data)
encoder=LabelEncoder()
df['Gender_encoded']=encoder.fit_transform(df['Gender'])
df
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
data={'Student': ['A', 'B', 'C'], 'Score': [55, 80, 65]}
df=pd.DataFrame(data)
sc=MinMaxScaler()
df['Normalized_Score']=sc.fit_transform(df[['Score']])
df
```

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
data={'Age': [22, 25, 47, 52, 46, 56, 23], 'Salary': [15000, 29000, 48000, 60000, 52000, 61000, 17000], 'Buy': [0, 0, 1, 1, 1, 1, 0]}
df=pd.DataFrame(data)
df
```

```python
x=df[['Age','Salary']]
y=df['Buy']
from sklearn.model_selection import train_test_split
(x_train,x_test,y_train,y_test)=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
svc=SVC()
svc.fit(x_train,y_train)
svc_pred=svc.predict(x_test)
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
rf_pred=rf.predict(x_test)
gb=GaussianNB()
gb.fit(x_train,y_train)
gb_pred=gb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
svc_accuracy=accuracy_score(y_test,svc_pred)
print("The SVC accuracy score is:",svc_accuracy)
print("The RandomForest score is:",accuracy_score(y_test,rf_pred))
print("The GaussianNB score is:",accuracy_score(y_test,rf_pred))
```
