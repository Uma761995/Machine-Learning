```python
import lightgbm
```

```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent=dataset[["charges"]]
```

```python
from sklearn.model_selection import train_test_split
(x_train,x_test,y_train,y_test)=train_test_split(independent, dependent, test_size=0.30, random_state=0) 
```

```python
from lightgbm import LGBMRegressor
regressor=LGBMRegressor(learning_rate=0.07,shrinkage_rate=0.1)
regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle
filename="finalized_model_creation_lgboost.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("finalized_model_creation_lgboost.sav",'rb'))
result=loaded_model.predict([[15,23.21,3,1,0]])
result
```
